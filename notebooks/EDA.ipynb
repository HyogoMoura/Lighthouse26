## 1. Conexão com o Postgres

```python

import pandas as pd
from sqlalchemy import create_engine

user="postgres"
password="postgres"
port=5433

engine = create_engine(f"postgresql://{user}:{password}@localhost:{port}/target_db")

df_clientes = pd.read_sql("SELECT * FROM clientes_crm", engine)
df_custos = pd.read_sql("SELECT * FROM custos_importacao", engine)
df_produtos = pd.read_sql("SELECT * FROM produtos_raw", engine)
df_vendas = pd.read_sql("SELECT * FROM vendas_2023_2024", engine)
```
